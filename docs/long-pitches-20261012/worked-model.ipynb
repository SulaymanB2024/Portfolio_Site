## Results

The cash figures are author sensitivities, not interchangeable company-reported metrics. Share dates, cash transformations and uncertainty are retained in each article and the input file.

```python
rows=['|Company|Entry USD|Underwriting cash $m|Base model USD|Rounded forecast USD|','|---|---:|---:|---:|---:|']
for company in inputs['companies']:
    data=output['companies'][company['symbol']]
    base=next(s for s in data['scenarios'] if s['name']=='Base')
    rows.append(f"|{company['symbol']}|{company['entry_close']:.2f}|{data['ttm']['owner_cash_proxy']:,.3f}|{base['terminal_price']:.2f}|{base['rounded_price_forecast']:.0f}|")
display(Markdown('\n'.join(rows)))
```
